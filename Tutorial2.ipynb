```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following
#1. Create a numpy array of size 2 x 2

import numpy as np

array_2x2 = np.array ([
   [1,2],
   [3,4]
])

print("This is my numpy array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,4))
zeroes_array = np.zeros((3,3))

print("Array of Ones :")
print(ones_array)

print("Array of Zeroes :")
print(zeroes_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it


random6x3 = np.random.rand(6,3)

print("Original 6x3 array:")
print(random6x3)

transposed = random6x3.T

print("\nTransposed array:")
print(transposed)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

n = 6
p = 4
m = 3

array1 = np.random.rand(n,p)
array2 = np.random.rand(p,m)

print("First array (n x p):")
print(array1)

print("Second array (p x m):")
print(array2)

dot_product = np.dot(array1, array2)

print("Dot product:")
print(dot_product)
```
